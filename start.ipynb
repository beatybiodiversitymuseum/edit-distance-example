# Filtering: Tokenizing, Stemming and Lemmatization

Text processing can involve multiple passses of cleaning. In this section, we look at ways to transform words into more computable forms.

## 1. Download the Natural Language Toolkit (nltk)

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

## 2. Tokenize: split words into chunks

```python
from nltk.tokenize import word_tokenize, sent_tokenize
text = "The cats are running faster than the dogs."
tokens = word_tokenize(text)
print(tokens)
```

## 3. Stem: get rid of things like suffixes

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed = [stemmer.stem(token) for token in tokens]
print(stemmed)
```

## 4. Remove stopwords

```python
from nltk.corpus import stopwords
import string
stop_words = set(stopwords.words("english"))

def allow(word):
    return word.lower() not in stop_words and word not in string.punctuation

filtered = [word for word in stemmed if allow(word)]
print(filtered)
```

## 5. Compare sentences
With the filtering applied, are these the same sentence?


```python
s1 = "The cats are running faster than the dogs."
s2 = "My cat can run faster than most dogs."

tokens_1 = word_tokenize(s1)
tokens_2 = word_tokenize(s2)

stemmed_1 = [stemmer.stem(token) for token in tokens_1]
stemmed_2 = [stemmer.stem(token) for token in tokens_2]

filtered_1 = [word for word in stemmed_1 if allow(word)]
filtered_2 = [word for word in stemmed_2 if allow(word)]

print(filtered_1)
print(filtered_2)
```

## 6. Part-of-Speech tagging
You can also filter based on part of speech.


```python
text = "The cats are running faster than the dogs."
tokens = word_tokenize(text)
tagged = nltk.pos_tag(tokens)
print(tagged)
```

```python
nouns = [word for (word, tag) in tagged if tag == "NNS"]
print(nouns)
```

## 6. Sort and add a column to compare values
