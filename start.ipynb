## 6. Sort and add a column to compare values

```python
df = df.sort_values(by=["Genus","Species"])
df["Previous Genus"] = df["Genus"].shift(1)
df["Previous Species"] = df["Species"].shift(1)
print(df)
```

## 7. Define the distance metric and apply

```python
def dist(row):
    if row["Genus"] == row["Previous Genus"] and row["Species"] == row["Previous Species"]:
        return 0
    else: 
        return 100
df["distance"] = df.apply(dist, axis=1)
print(df.sort_values(by=["distance"]))
df.to_excel("output.xlsx", index=False)
```

## 8. Import a pre-defined distance metric

```python
from Levenshtein import distance
def dist(row):
    if row["Genus"] == row["Previous Genus"]:
        return distance(row["Species"], row["Previous Species"])
    return 100
df["distance"] = df.apply(dist, axis=1)
df = df.sort_values(by=["distance"])
df.to_excel("output_levenshtein.xlsx", index=False)
```
